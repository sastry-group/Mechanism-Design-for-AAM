```python
import subprocess
import os
from itertools import product

main_script_path = os.path.join(os.path.abspath(''), 'main.py')
file_list = ["../test_cases/case0.json"]

lambdas = [0.5]
parameter_combinations = list(product(lambdas))

idx = 0
for file in file_list:
    args = [
                "python", main_script_path,
                "--file", file,
                "--method", "vcg",
                "--output_bsky", str(False),
                "--run_up_to_auction", str(0),
                "--force_overwrite",
                "--save_pkl_files", "True",
            ]    

    print(f"Running configuration {idx + 1}/{len(parameter_combinations)}: ")
    try:
        subprocess.run(args, check=True)
    except subprocess.CalledProcessError as e:
        print(f"Configuration {idx + 1} failed: {e}. Skipping to the next configuration.")
    idx += 1
```
